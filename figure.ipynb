```python
# Import and Load the Assignment 2 Results
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import os
os.makedirs("figures", exist_ok=True)
sns.set_theme(style="whitegrid", context="talk")
```

```python
# Load the Assignment 2 results 
df = pd.read_csv("C:\Users\User\OneDrive\Desktop\WLD_2024_FINDEX_v02_M_CSV\WLD_2024_FINDEX_v02_M_CSV\findex_microdata_2025_labelled_update112425-For description\findex_microdata_2025_labelled_update112425.csv")
df.head()
```

```python
# Create Chart 1 (Labeled)
plt.figure(figsize=(8, 5))
ax = sns.barplot(data=df, x="category_column", y="value_column", palette="viridis")
ax.set_title("Figure 1: Mean Value by Category")
ax.set_xlabel("Category")
ax.set_ylabel("Mean Value")
plt.tight_layout()
plt.savefig("figures/figure1_bar.png", dpi=300)
plt.show()
```

```python
# Create Chart 2 (Labeled)
plt.figure(figsize=(8, 5))
ax = sns.lineplot(data=df, x="x_column", y="y_column", hue="group_column", marker="o")
ax.set_title("Figure 2: Trend of Y Over X by Group")
ax.set_xlabel("X Axis Label (units)")
ax.set_ylabel("Y Axis Label (units)")
plt.legend(title="Group")
plt.tight_layout()
plt.savefig("figures/figure2_line.png", dpi=300)
plt.show()
```
